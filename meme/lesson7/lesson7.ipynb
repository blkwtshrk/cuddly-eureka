```python
while True: 
    user_input = input('enter sth: ')
    if user_input == 'exit':
        break
    elif user_input == 'skip':
        print('wasting...')
        continue
    elif len(user_input) > 10:
        print('your input is too long')
    else: 
        print('input is ok')

print('good bye')
```

```python
text = 'I am Malenia, Blade of Miquella… and I have never known defeat.'

words = text.split()
fin_words = []
for word in words:
    if 'o' in word:
        print(word)
        continue
    fin_words.append(word)

print(' '.join(fin_words))
```

```python
text = 'I am Malenia, Blade of Miquella… and I have never known defeat.'

words = text.split()
fin_words = []
for word in words:
    if word == 'and':
        break
    elif 'o' in word:
        print(word)
        continue
    fin_words.append(word)

print(' '.join(fin_words))
```

### Functions

Самые основные программные структуры в языке Python, которые позволяют многократное испольнование кода и уменьшающее его избыточность. Также позволяют разбить сложную систему на достаточно простые и легко управляемые части. Они могут принимать аргументы и возвращать результаты выполнения.

[DRY] - принцип в программировании, который расшифровывается как: "Не повторяйся"
D - don't
R - repeat
Y - yourself

```python
# пример хуевости:
price = 100
final_price = price - price * 0.1

order_price = 200
order_final = order_price - order_price * 0.1

print(final_price)
print(order_final)
```

```python
#пример охуенности
def apply_discount(price):
    return price - price * 0.1

final_price = apply_discount(100)
order_final = apply_discount(200)
order_prefinal = apply_discount(300)

print(final_price)
print(order_final)
print(order_prefinal)
```

```python
def add(a, b):  # a и b - параметры
    return a + b

result = add(2, 3)  # 2 и 3 - аргументы
print(result)

# функцию можно представить как форму:
# параметры -> поля формы (имя, достижения)
# аргументы ->  то, что ты в них вводишь ("Илюха", "долбоеб")
```
